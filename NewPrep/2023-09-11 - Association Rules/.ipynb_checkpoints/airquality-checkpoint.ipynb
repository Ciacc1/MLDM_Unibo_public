### Read the dataset, show its head, shape and description


```python
# Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Variables
file_name= 'AirQualityUCI.csv'
random_state = 42
np.random.seed(random_state)

df = pd.read_csv(file_name, sep=';')
df.head()
```

```python
df.shape
```

```python
df.describe()
```

### Eliminate totally null columns and totally null rows, eliminate columns with less than 1/3 of non null values; fill the remaining NaN values with the mean of the column

```python
dfclean=df.dropna(how='all')
dfclean=dfclean.dropna(thresh=(len(dfclean)*1/3),axis=1)
dfclean.shape
print(f"NaN rimanenti dopo riempimento: {dfclean.isnull().sum().sum()}")
```

```python

#Filling remaining NaN values with the mean of the column
numericcols = dfclean.select_dtypes('number').columns
dfclean[numericcols] = dfclean[numericcols].fillna(dfclean[numericcols].mean())
```

```python
dfclean.shape
```

```python
dfclean.head()
```

```python
print(f"NaN rimanenti dopo riempimento: {dfclean.isnull().sum().sum()}")
```

### Drop Time, convert Date from string to datetime and group by Date using mean as aggregate function

```python
from datetime import date
dffinal=dfclean.drop('Time',axis=1)
#Converting 'Date' from string to datetime
dffinal['Date'] = pd.to_datetime(dffinal['Date'],format="%d/%m/%Y")
dffinal.head()
```

```python
df_grouped = dffinal.groupby('Date').mean()
df_grouped.head()
```

### Discretise continuous values with two bins, kmeans strategy and onehot-dense encoding

```python

from sklearn.preprocessing import KBinsDiscretizer
#Discretizing continuous values with two bins, kmeans strategy and onehot-dense encoding
kbd = KBinsDiscretizer(n_bins=2,strategy='kmeans',encode='onehot-dense').set_output(transform='pandas')
dftransformed = kbd.fit_transform(df_grouped)
dftransformed.head()
```

###  Discretization/encoding generates 0/1 values; convert the binary values obtained into boolean, as requested by Apriori

```python
dfapriori=dftransformed.astype(bool)
dfapriori.head()
```

###  Set the names of two columns generated by the discretisation of each attribute A to A_low, A_high (with discretisation/one-hot-encoding, each original column generates two columns, the first is for the low values, the second for the high values)

```python
newcolnames = {}
for column in dftransformed.columns :
    colname = column[:3]
    if column.endswith('_0.0') : newcolnames[column] = colname + '_low'
    elif column.endswith('_1.0') : newcolnames[column] = colname + '_high'
    else : newcolnames[column] = column

dfrenamed = dftransformed.rename(columns=newcolnames)
dfrenamed.head()
```

### find a value of min_support such that the Apriori algorithm generates at least 8 frequent itemsets with at least 2 items, output the result

```python
min_itemsets = 8
min_item_in_itemset = 2
# "Reasonable" range
support_range = list(np.arange(0.1, 0.001, -0.001))

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
min_support = 0
for s_value in support_range:
    print(f"Trying support value {s_value:.2f}")
    frequent_itemsets = apriori(dfrenamed, min_support = s_value, use_colnames = True)
    # Calculate the number of itemsets that contain at least `min_item_in_itemset` items
    # frequent_itemsets must contains al least min_itemsets itemsset , and every item set must contains al least min_item_in_itemset item
    itemsets_above_threshold = sum([len(itemset) >= min_item_in_itemset for itemset in frequent_itemsets.itemsets])
    if itemsets_above_threshold >= min_itemsets:
        min_support = s_value
        break
if min_support == 0:
    print("No itemset found! Try again with a bigger range!")
else:
    print(f"I've selected min_support = {min_support:.3f}, which produced , {len(frequent_itemsets)} itemsets, {itemsets_above_threshold} of which had more than {min_item_in_itemset} items")
```

### find the minimum metric threshold such that at least 100 association rules are extracted from the frequent itemsets found and show the metrics used and the best 10 rules by descending confidence and support

```python
metric_threshold_range = np.arange(1, 0.01, -0.01)
min_association_rule = 100
min_metric_threshold = 0
association_rule_found = 0
current_metric = "lift"

for metric_value in metric_threshold_range:

    rules = association_rules(frequent_itemsets, metric=current_metric, min_threshold=metric_value)
    if rules.shape[0] >= min_association_rule:
        association_rule_found = rules.shape[0]
        min_metric_threshold = metric_value
        break

if association_rule_found == 0:
    print("No association rule! Try again with a bigger range!")
else:
    print(f"I've selected metric {current_metric}  with metric_value = {metric_value:.2f}, which produced , {association_rule_found} association_rules")
```

```python
# Print the first n rules found, sorted by descending confidence and support
n_rule = 10
sorted_rules=rules.sort_values(by=['confidence','support'],ascending=False).reset_index(drop=True)
sorted_rules[:10]
```
