```python
#Imports and variables definition
import pandas as pd
import numpy as np

from sklearn.preprocessing import KBinsDiscretizer
from mlxtend.frequent_patterns import apriori, association_rules

random_state = 42
np.random.seed(random_state)
```

### 1. Read the dataset, show its head, shape and description

```python
#The dataset contains an header row and uses ';' as separator
url = './AirQualityUCI.csv'
sep = ';'

df = pd.read_csv(url,sep=sep)
df.head() #Showing head
```

```python
#Showing description
df.describe()
```

### 2. Eliminate totally null columns and totally null rows, eliminate columns with less than 1/3 of non null values; fill the remaining NaN values with the mean of the column

```python
#Eliminating totally null columns
df = df.dropna(how='all',axis=1)

#Eliminating totally null rows
df = df.dropna(how='all',axis=0)

#Eliminating columns with less than 1/3 of non null values
columnstodrop = []
for column in df.columns :
    if sum(df[column].isna()) >= len(df[column])/3 : #If at least 1/3 of the values is null
        columnstodrop.append(column)

df = df.drop(columns=columnstodrop,axis=1)

#Filling remaining NaN values with the mean of the column
numericcols = df.select_dtypes('number').columns
df[numericcols] = df[numericcols].fillna(df[numericcols].mean())

df.describe()
```

### 3. Drop `Time`, convert `Date` from string to datetime and group by `Date` using mean as aggregate function

```python
#Dropping 'Time'
df = df.drop(columns='Time')
```

```python
#Converting 'Date' from string to datetime
df['Date'] = pd.to_datetime(df['Date'],format="%d/%m/%Y")
```

```python
#Grouping by date using mean as aggregate function
dfgrouped = df.groupby(by='Date').mean()
dfgrouped.head()
```

### 4. Preparation of the boolean matrix
- Discretise continuous values with two bins, kmeans strategy and onehot-dense encoding
- Discretization/encoding generates 0/1 values; convert the binary values obtained into boolean, as requested by Apriori

```python
#Discretizing continuous values with two bins, kmeans strategy and onehot-dense encoding
kbd = KBinsDiscretizer(n_bins=2,strategy='kmeans',encode='onehot-dense').set_output(transform='pandas')
dftransformed = kbd.fit_transform(dfgrouped)
dftransformed.head()
```

```python
#Converting binary values obtained into boolean
dftransformed = dftransformed.astype(bool)
dftransformed.head()
```

### 5. Set the names of two columns generated by the discretisation of each attribute A to A_low, A_high (with discretisation/one-hot-encoding, each original column generates two columns, the first is for the low values, the second for the high values)

```python
newcolnames = {}
for column in dftransformed.columns :
    colname = column[:3]
    if column.endswith('_0.0') : newcolnames[column] = colname + '_low'
    elif column.endswith('_1.0') : newcolnames[column] = colname + '_high'
    else : newcolnames[column] = column

dfrenamed = dftransformed.rename(columns=newcolnames)
dfrenamed.head()
```

### 6. find a value of min_support such that the Apriori algorithm generates at least 8 frequent itemsets with at least 2 items, output the result

```python
#Requirements
min_itemsets = 8
min_items_in_itemsets = 2

#Creating a list of min_support values to try the apriori algorithm with
min_support_list = list(np.arange(0.1,0.001,-0.001))

min_support = 0
for min_sup in min_support_list :
    frequentitemsets = apriori(dfrenamed, min_support=min_sup, use_colnames=True)

    #Check if every itemset has at least min_items_in_itemsets items
    itemsetsabovethreshold = sum([len(itemset) >= min_items_in_itemsets for itemset in frequentitemsets.itemsets])

    if itemsetsabovethreshold > min_itemsets :
        min_support = min_sup
        break

if min_support == 0 : print("No min_support value satisfies the requirements")
else : print(f"min_support = {min_support} generated {itemsetsabovethreshold} itemsets which satisfy the requirements")
```

### 7. find the minimum metric threshold such that at least 100 association rules are extracted from the frequent itemsets found and show the metrics used and the best 10 rules by descending confidence and support

```python
#Requirements
min_association_rules = 100

#Creating a list of min_threshold values to try the association_rules algorithm with
min_threshold_list = list(np.arange(1.0,0.01,-0.01))
metric = 'lift'

min_threshold = 0
for min_thres in min_threshold_list : 
    rules = association_rules(frequentitemsets,metric=metric,min_threshold=min_thres)

    if len(rules) >= min_association_rules :
        min_threshold = min_thres
        break

if min_support == 0 : print("No min_threshold value satisfies the requirements")
else : print(f"min_threshold = {min_threshold} generated {len(rules)} rules using metric: {metric}")
```

```python
#Showing best 10 rules by descending confidence and support
rules.sort_values(by=['confidence','support'],ascending=[False,False]).head(10)
```
